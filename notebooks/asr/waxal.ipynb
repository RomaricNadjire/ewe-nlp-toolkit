# WaxalNLP — Chargement de l'éwé uniquement

Les trois jeux de données (`google/WaxalNLP`, `novelwolde36/WaxalNLP`, `Phsntom/WaxalNLP`) sont des variantes du corpus **WaxalNLP**, organisé **par configuration de langue** (ex. `sna_asr` = Shona, `ewe_asr` = éwé). Il n'existe pas de configuration « toutes langues » : **charger la config `ewe_asr` revient donc à filtrer pour ne garder que l'éwé.**

Splits de `ewe_asr` (par dataset) :
- `train` ≈ 15 000 ex. (~4,5 Go), `validation` / `test` ≈ 1 900 ex.
- `unlabeled` ≈ 184 000 ex. (~55 Go) → **ignoré par défaut** (ajoutez `"unlabeled"` à `SPLITS` si nécessaire).

Notes :
- `ewe_tts` (synthèse vocale) n'existe que sur `google/WaxalNLP`.
- On passe `columns=` à `load_dataset` pour contourner un bug du dataset (colonne pandas résiduelle `__index_level_0__` qui provoque sinon une `CastError`).
- En cas d'accès restreint : `huggingface-cli login`.


```python
# (Optionnel) Aperçu rapide SANS tout télécharger, via le streaming.
# Permet de vérifier que la config éwé est correcte avant le téléchargement complet.
from datasets import load_dataset

# 'columns=' contourne un bug du dataset : les parquets contiennent une colonne
# pandas résiduelle '__index_level_0__' absente du schéma déclaré (sinon CastError).
COLS = ["id", "speaker_id", "transcription", "language", "gender", "audio"]

preview = load_dataset("google/WaxalNLP", "ewe_asr", split="train",
                       streaming=True, columns=COLS)
preview = preview.select_columns(["language", "transcription"])  # évite de décoder l'audio
for ex in preview.take(3):
    print(ex["language"], "|", ex["transcription"][:80])
```

```python
from datasets import DatasetDict, load_dataset

# WaxalNLP est organisé par configuration de langue : "ewe_asr" = éwé.
# Charger cette config revient donc à filtrer pour ne garder QUE l'éwé.
DATASETS = ["google/WaxalNLP", "novelwolde36/WaxalNLP", "Phsntom/WaxalNLP"]
EWE_CONFIG = "ewe_asr"
SPLITS = ["train", "validation", "test"]   # "unlabeled" (~55 Go/dataset) ignoré

# 'columns=' contourne un bug du dataset : les parquets contiennent une colonne
# pandas résiduelle '__index_level_0__' absente du schéma déclaré (sinon CastError).
COLS = ["id", "speaker_id", "transcription", "language", "gender", "audio"]

# On charge chaque split séparément pour ne PAS télécharger le split "unlabeled".
ewe = {}
for name in DATASETS:
    dd = DatasetDict({
        s: load_dataset(name, EWE_CONFIG, split=s, columns=COLS) for s in SPLITS
    })
    ewe[name] = dd
    print(name, "->", {s: len(dd[s]) for s in SPLITS})

ewe
```

```python
# Vérifie que seules des données en éwé sont présentes
# (la colonne 'language' est lue sans décoder l'audio -> rapide)
for name, dd in ewe.items():
    langues = set()
    for s in dd:
        langues |= set(dd[s]["language"])
    print(name, "-> langues =", langues)

# Exemple d'un enregistrement
ex = ewe["google/WaxalNLP"]["train"][0]
print("\nTranscription :", ex["transcription"])
print("Langue        :", ex["language"])
print("Sampling rate :", ex["audio"]["sampling_rate"])
print("Audio shape   :", ex["audio"]["array"].shape)
```
